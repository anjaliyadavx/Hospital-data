## Data Visualization

```python
df
```

```python
df['Blood Type']=df['Blood Type'].astype(str).str.strip().str.upper()
```

```python
df['Blood Type']= df['Blood Type'].str.replace(r"(\+|-)+$",lambda x: x.group(0)[0].upper(),regex=True)
```

```python
df['Blood Type'].value_counts()
```

```python
df
```

```python
df['Gender'].value_counts().plot(kind='bar', color=['blue', 'pink'])
plt.title("Gender Distribution")
plt.xlabel("Gender")
plt.ylabel("Count")
plt.show()
```

```python
blood_type_counts = df['Blood Type'].value_counts()
colors = []
for bt in blood_type_counts.index:
    if bt.startswith('A'):
        colors.append('red')
    elif bt.startswith('B'):
        colors.append('maroon')
    elif bt.startswith('O'):
        colors.append('lightcoral')
    elif bt.startswith('AB'):
        colors.append('purple')
    else:
        colors.append('grey')

blood_type_counts.plot(kind='bar', figsize=(10,5), color=colors)
plt.title("Blood Type Distribution")
plt.xlabel("Blood Type")
plt.ylabel("Count")
plt.xticks(rotation=45)
plt.show()
```

```python
df['Medical Condition'].value_counts().plot(kind='bar',figsize=(10,6),color='green')
plt.title("Medical Conditions")
plt.xlabel("Condition")
plt.ylabel("Patients")
plt.xticks(rotation=45)
plt.show()
```

```python
df['Admission Type'].value_counts().plot(kind='pie', autopct='%1.1f%%')
plt.title("Admission Type")
plt.ylabel("")
plt.show()
```

```python
df['Insurance Provider'].value_counts().plot(kind='bar')
plt.title("Insurance Providers")
plt.xlabel("Provider")
plt.ylabel("Patients")
plt.xticks(rotation=45)
plt.show()
```

```python
df['Medication'].value_counts().plot(kind='bar',color='gold')
plt.title("Medication Usage")
plt.xlabel("Medication")
plt.ylabel("Count")
plt.xticks(rotation=45)
plt.show()
```

```python
test_result_counts = df['Test Results'].value_counts()
colors = []
for bt in test_result_counts.index:
    if bt == 'Unknown':
        colors.append('yellow')
    elif bt == 'Abnormal':
        colors.append('brown')
    else:
        colors.append('grey')
test_result_counts.plot(kind='bar', color=colors)
plt.title("Test Results")
plt.xlabel("Result")
plt.ylabel("Patients")
plt.show()
```

```python
plt.hist(df['Age'], bins=10,color='olive', edgecolor='green')
plt.title("Age Distribution")
plt.xlabel("Age")
plt.ylabel("Patients")
plt.show()
```

```python
plt.hist(df['Billing Amount'], bins=15, color='red', edgecolor='black')
plt.title("Billing Amount Distribution")
plt.xlabel("Billing Amount")
plt.ylabel("Patients")
plt.show()
```

```python
df.groupby('Admission Type')['Billing Amount'].mean().plot(kind='bar',color='purple')
plt.title("Average Billing by Admission Type")
plt.xlabel("Admission Type")
plt.ylabel("Average Billing")
plt.show()
```
